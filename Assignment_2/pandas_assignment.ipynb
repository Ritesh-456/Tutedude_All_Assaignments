### 1. Create a DataFrame
I need to make a dataframe with Name, Age, Department, and Salary.
Let me try using a dictionary first.

```python
my_data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

# creating the dataframe
df = pd.DataFrame(my_data)
# let's see if it worked
print(df)
```

#### a. Print the first five rows of the DataFrame
I remember there is a function called `.head()` for this. 

```python
# using head() to show just the top 5 rows
first_five = df.head()
print(first_five)
```

#### b. Get the summary statistics of the 'Age' and 'Salary' columns
I think the function is `describe()`. Let me try it on the whole dataframe first to see what it does.

```python
# trying to describe everything first
# print(df.describe()) # This gave me stats for age and salary automatically!

# But the question asks specifically for Age and Salary columns.
# I need to select them as a list inside the brackets
stats = df[['Age', 'Salary']].describe()
print(stats)
```

#### c. Calculate the average salary of employees in the 'HR' department
This one is tricky. First I need to find only the HR employees, then get their salary, and find the mean.
Let me test finding HR employees first.

```python
# checking my logic to filter HR
# hr_only = df[df['Department'] == 'HR']
# print(hr_only) # It works! It shows Alice and Eve.

# Now I'll get the salary column from that and calculate the average
hr_employees = df[df['Department'] == 'HR']
# extracting the salary column
hr_salaries = hr_employees['Salary']
# calculating the mean
average_hr_salary = hr_salaries.mean()

print("The average salary for HR is:", average_hr_salary)
```

### 2. Add a new column, 'Bonus', which is 10% of the salary
I can just multiply the salary column by 0.10 and assign it to a new column name.

```python
# adding a new column by doing math on an existing one
df['Bonus'] = df['Salary'] * 0.10

# printing to check if Bonus is there now
print(df)
```

### 3. Filter the DataFrame to show employees aged between 25 and 30
I can use the `>=` and `<=` operators, or maybe there's a specific function. I'll try using the `between()` function because it seems easier to read.

```python
# filtering for age between 25 and 30 inclusive
# I'll save it to a new variable so I don't mess up the original dataframe
filtered_employees = df[df['Age'].between(25, 30)]
print("Employees between 25 and 30:")
print(filtered_employees)
```

### 4. Group the data by 'Department' and calculate the average salary for each department
I will use the `groupby` function for this. Grouping by department, then selecting salary, then taking the mean.

```python
# grouping by department
grouped = df.groupby('Department')
# taking the mean of the salary
average_salaries_by_dept = grouped['Salary'].mean()

print(average_salaries_by_dept)
```

### 5. Sort the DataFrame by 'Salary' in ascending order and save the result to a new CSV file
Sorting is done with `sort_values`. Ascending is usually the default, but I'll specify it just to be safe.
Then I need to figure out how to export to CSV.

```python
# sorting the dataframe
sorted_df = df.sort_values(by='Salary', ascending=True)
print("Sorted DataFrame:")
print(sorted_df)

# saving to CSV. I'll use index=False so it doesn't save the row numbers (0, 1, 2...) into the file
sorted_df.to_csv('sorted_employees.csv', index=False)
print("Successfully saved to sorted_employees.csv!")
```
