## Working with Pandas

**Objective:** Learn to create and manipulate dataframes for data analysis.


```python
import pandas as pd
```

### Task 1: DataFrame Initialization & Baseline Statistics
We instantiate a standard dictionary containing employee demographics and parse it into a pandas DataFrame object to compute profile summary statistics.


```python
# Initialize data dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

# Create DataFrame
df = pd.DataFrame(data)

# 1.a Print first five rows
print("First Five Rows:")
print(df.head())

# 1.b Summary statistics for 'Age' and 'Salary'
print("\nSummary Statistics:")
print(df[['Age', 'Salary']].describe())

# 1.c Average salary of employees in 'HR'
hr_avg_salary = df[df['Department'] == 'HR']['Salary'].mean()
print(f"\nAverage Salary in HR Department: {hr_avg_salary}")
```

### Tasks 2 - 5: Data Transformation, Conditional Filtering, and Serialization
Here, we perform feature extraction by adding a calculated bonus column, subset records based on row filter criteria, compute aggregates by groups, and sort findings to a physical output spreadsheet.


```python
# 2. Add 'Bonus' column (10% of salary)
df['Bonus'] = df['Salary'] * 0.10
print("DataFrame after adding Bonus column:")
print(df)
print("\n" + "="*50 + "\n")

# 3. Filter DataFrame for employees aged between 25 and 30 (inclusive)
filtered_df = df[(df['Age'] >= 25) & (df['Age'] <= 30)]
print("Employees aged between 25 and 30:")
print(filtered_df)
print("\n" + "="*50 + "\n")

# 4. Group by Department and find average salary
dept_avg_salary = df.groupby('Department')['Salary'].mean().reset_index()
print("Average Salary per Department:")
print(dept_avg_salary)
print("\n" + "="*50 + "\n")

# 5. Sort by Salary ascending and export to a new CSV file
sorted_df = df.sort_values(by='Salary', ascending=True)
sorted_df.to_csv('sorted_employees.csv', index=False)
print("Sorted DataFrame saved to 'sorted_employees.csv'. Contents:")
print(sorted_df)
```
