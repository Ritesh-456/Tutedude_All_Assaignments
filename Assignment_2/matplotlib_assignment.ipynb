## Working with Matplotlib

**Objective:** Practice data visualization techniques for better data representation.


```python
import matplotlib.pyplot as plt
import numpy as np
```

### Task 1: Basic Trend Visualization using Line Plot
We build a trend line representing a linear continuous timeline distribution coordinate map. Labels and grid metrics are overlayed onto the canvas layout.


```python
# Data coordinates
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]

# Plot generation
plt.figure(figsize=(6, 4))
plt.plot(x, y, marker='o', linestyle='-', color='b', label='Growth Trend')

# Customizations
plt.title('Line Plot of Sample Growth Trend')
plt.xlabel('X-Axis Scale Units')
plt.ylabel('Y-Axis Value Metrics')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Display plot
plt.show()
```

### Task 2: Comparative Categorical Measurement using Bar Chart
We map out categorical data variables evaluating grades scored by structural user categories using varied bar canvas coloring styles.


```python
# Categorical metrics
students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]

# Bar chart configuration
plt.figure(figsize=(6, 4))
colors = ['skyblue', 'salmon', 'lightgreen', 'gold']
plt.bar(students, marks, color=colors, edgecolor='grey')

# Customizations
plt.title('Student Academic Subject Performance Breakdown')
plt.xlabel('Student Identification Names')
plt.ylabel('Evaluated Test Scores / Marks')
plt.ylim(0, 100)

# Display plot
plt.show()
```

### Task 3: Distribution Share Analysis using Pie Chart
We track structural percentage slices belonging to core geo revenue vectors. The sector maintaining the absolute peak value is pulled away visually using localized data wedge separation settings.


```python
# Share metrics
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]

# Highlight index 0 (North America) because it holds highest value (45)
explode = (0.1, 0, 0, 0)  

# Pie Chart construction
plt.figure(figsize=(5, 5))
plt.pie(revenue, explode=explode, labels=regions, autopct='%1.1f%%',
        startangle=140, colors=['teal', 'orange', 'green', 'magenta'])

# Title mapping
plt.title('Global Corporate Revenue Percentage Distribution')

# Display plot
plt.show()
```

### Task 4: Frequency Probability Distribution using Histogram
We build a continuous statistical layout tracking observations sampled from random vector integers across uniform parameters.


```python
# Initialize data arrays
np.random.seed(42)
random_integers = np.random.randint(1, 101, size=1000)

# Build Histogram visualization
plt.figure(figsize=(7, 4))
plt.hist(random_integers, bins=20, color='purple', edgecolor='black', alpha=0.7)

# Overlay features
plt.title('Frequency Distribution Histogram (N=1000)')
plt.xlabel('Random Value Class Ranges (1-100)')
plt.ylabel('Observed Instance Frequency Count')
plt.grid(axis='y', linestyle=':', alpha=0.5)

# Display plot
plt.show()
```
