```python
import numpy as np
```

1. Create a 1D NumPy array with integers from 1 to 20. Perform the following operations:

```python
arr = np.arange(1,21)
arr
```

a. Calculate the sum, mean, median, and standard deviation of the elements in the array.

```python
print(f"Sum: {np.sum(arr)}")
print(f"Mean: {np.mean(arr)}")
print(f"Median: {np.median(arr)}")
print(f"standard deviation: {np.std(arr)}")
```

b. Find the indices of elements greater than 10 in the array.

```python
indices = np.where(arr > 10)
print(indices)
```

2. Create a 2D NumPy array of shape 4 X 4 with numbers ranging from 1 to 16

a. Print the array.

```python
arr_2d = np.arange(1,17).reshape(4,4)
arr_2d
```

b. Find the transpose of the array.

```python
print(f"Transpose of arr_2d:\n{np.transpose(arr_2d)}")
```

c. Calculate the row-wise and column-wise sums of the array.

```python
print(f"Row wise sum\n{np.sum(arr_2d, axis=1)}")
print(f"Column wise sum\n{np.sum(arr_2d, axis=0)}")
```

3. Create two 3 X 3 arrays filled with random integers between 1 and 20.

```python
array1 = np.random.randint(1, 21, size=(3, 3))
array2 = np.random.randint(1, 21, size=(3, 3))

print("Array 1:\n", array1)
print("\nArray 2:\n", array2)
```

a. Perform element-wise addition, subtraction, and multiplication.

```python
addition = array1 + array2
subtraction = array1 - array2
multiplication = array1 * array2

print("Addition:\n", addition)
print("\nSubtraction:\n", subtraction)
print("\nMultiplication:\n", multiplication)
```

b. Compute the dot product of the two arrays.

```python
dot_product = np.dot(array1, array2) 

print("Dot Product:\n", dot_product)
```

4. Reshape a 1D array of size 12 into a 3 X 4 2D array and slice the first two rows
and last two columns.

```python
arr_1d = np.arange(1, 13)
arr_reshaped = arr_1d.reshape(3, 4)
print("Reshaped Array:\n", arr_reshaped)

sliced_arr = arr_reshaped[:2, -2:]
print("\nSliced Array:\n", sliced_arr)
```
