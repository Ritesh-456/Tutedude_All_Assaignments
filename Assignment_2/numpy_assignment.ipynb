## Working with NumPy

**Objective:** Understand the basics of NumPy, array creation, and manipulation.

```python
import numpy as np
```

### Task 1: 1D Array Operations
We create a 1D NumPy array containing integers from 1 to 20. Then we compute descriptive statistics (sum, mean, median, standard deviation) and extract indices where elements exceed 10.


```python
# Create a 1D array from 1 to 20
arr_1d = np.arange(1, 21)
print("Original 1D Array:", arr_1d)

# 1.a Calculate statistics
arr_sum = np.sum(arr_1d)
arr_mean = np.mean(arr_1d)
arr_median = np.median(arr_1d)
arr_std = np.std(arr_1d)

print(f"Sum: {arr_sum}")
print(f"Mean: {arr_mean}")
print(f"Median: {arr_median}")
print(f"Standard Deviation: {arr_std:.4f}")

# 1.b Find indices of elements greater than 10
indices_gt_10 = np.where(arr_1d > 10)[0]
print("Indices of elements > 10:", indices_gt_10)
```

### Task 2: 2D Array Matrix Operations
We construct a $4 \times 4$ matrix ranging from 1 to 16, compute its transpose, and calculate row-wise and column-wise aggregations.


```python
# Create a 4x4 2D array
arr_2d = np.arange(1, 17).reshape(4, 4)
print("Original 2D Array:\n", arr_2d)

# 2.b Transpose of the array
arr_transpose = arr_2d.T
print("\nTransposed Array:\n", arr_transpose)

# 2.c Row-wise and column-wise sums
row_sums = np.sum(arr_2d, axis=1)
col_sums = np.sum(arr_2d, axis=0)

print("\nRow-wise sums:", row_sums)
print("Column-wise sums:", col_sums)
```

### Task 3: Operations on Random Arrays
We generate two distinct $3 \times 3$ matrices populated with random integers between 1 and 20. We perform element-wise addition, subtraction, multiplication, and matrix dot product multiplication.


```python
# Set seed for reproducibility
np.random.seed(42)

# Create two 3x3 random arrays between 1 and 20
matrix_A = np.random.randint(1, 21, size=(3, 3))
matrix_B = np.random.randint(1, 21, size=(3, 3))

print("Matrix A:\n", matrix_A)
print("\nMatrix B:\n", matrix_B)

# 3.a Element-wise operations
print("\nElement-wise Addition:\n", matrix_A + matrix_B)
print("\nElement-wise Subtraction:\n", matrix_A - matrix_B)
print("\nElement-wise Multiplication:\n", matrix_A * matrix_B)

# 3.b Dot product
dot_product = np.dot(matrix_A, matrix_B)
print("\nDot Product Matrix multiplication:\n", dot_product)
```

### Task 4: Reshaping and Slicing
We initialize a 1D array of 12 elements, reconstruct it into a $3 \times 4$ grid shape, and filter specific rows and columns using multi-axis slicing syntax.


```python
# Create 1D array of size 12 and reshape to 3x4
base_arr = np.arange(1, 13)
reshaped_arr = base_arr.reshape(3, 4)
print("Reshaped 3x4 Array:\n", reshaped_arr)

# Slice the first two rows and last two columns
# Rows: 0, 1 (up to index 2 exclusive)
# Columns: last two (-2 to end)
sliced_arr = reshaped_arr[0:2, -2:]
print("\nSliced Sub-array (First 2 rows, Last 2 columns):\n", sliced_arr)
```
